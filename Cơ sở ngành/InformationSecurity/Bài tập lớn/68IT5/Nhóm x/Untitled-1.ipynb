```python
from Crypto.Cipher import DES
from Crypto.Util.Padding import pad, unpad

def des_encrypt(plain_text, key):
    cipher = DES.new(key, DES.MODE_ECB)
    padded_text = pad(plain_text.encode(), DES.block_size)
    encrypted = cipher.encrypt(padded_text)
    return encrypted

def des_decrypt(cipher_text, key):
    cipher = DES.new(key, DES.MODE_ECB)
    decrypted = unpad(cipher.decrypt(cipher_text), DES.block_size)
    return decrypted.decode()

# Ví dụ
key = b'8bytekey'  # Đúng 8 byte
data = "Hello DES"
enc = des_encrypt(data, key)
dec = des_decrypt(enc, key)
print("Encrypted:", enc)
print("Decrypted:", dec)
```

```python
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad, unpad
from Crypto.Random import get_random_bytes

def aes_encrypt(plain_text, key):
    cipher = AES.new(key, AES.MODE_CBC)
    ct_bytes = cipher.encrypt(pad(plain_text.encode(), AES.block_size))
    return cipher.iv, ct_bytes

def aes_decrypt(cipher_text, key, iv):
    cipher = AES.new(key, AES.MODE_CBC, iv)
    pt = unpad(cipher.decrypt(cipher_text), AES.block_size)
    return pt.decode()

# Ví dụ
key = get_random_bytes(16)  # 16 byte cho AES-128
data = "Hello AES"
iv, enc = aes_encrypt(data, key)
dec = aes_decrypt(enc, key, iv)
print("Encrypted:", enc)
print("Decrypted:", dec)
```

```python
from Crypto.PublicKey import RSA
from Crypto.Cipher import PKCS1_OAEP

# Tạo cặp khóa
key = RSA.generate(2048)
private_key = key
public_key = key.publickey()

def rsa_encrypt(plain_text, public_key):
    cipher = PKCS1_OAEP.new(public_key)
    return cipher.encrypt(plain_text.encode())

def rsa_decrypt(cipher_text, private_key):
    cipher = PKCS1_OAEP.new(private_key)
    return cipher.decrypt(cipher_text).decode()

# Ví dụ
data = "Hello RSA"
enc = rsa_encrypt(data, public_key)
dec = rsa_decrypt(enc, private_key)
print("Encrypted:", enc)
print("Decrypted:", dec)
```

```python
from Crypto.Cipher import DES, AES, PKCS1_OAEP
from Crypto.PublicKey import RSA
from Crypto.Util.Padding import pad, unpad
from Crypto.Random import get_random_bytes
import base64

# ---------- DES ----------
def des_encrypt(plain_text, key):
    cipher = DES.new(key, DES.MODE_ECB)
    padded = pad(plain_text.encode(), DES.block_size)
    encrypted = cipher.encrypt(padded)
    return base64.b64encode(encrypted).decode()

def des_decrypt(cipher_b64, key):
    cipher = DES.new(key, DES.MODE_ECB)
    decrypted = cipher.decrypt(base64.b64decode(cipher_b64))
    return unpad(decrypted, DES.block_size).decode()

# ---------- AES ----------
def aes_encrypt(plain_text, key):
    cipher = AES.new(key, AES.MODE_CBC)
    ct = cipher.encrypt(pad(plain_text.encode(), AES.block_size))
    return base64.b64encode(cipher.iv + ct).decode()

def aes_decrypt(cipher_b64, key):
    raw = base64.b64decode(cipher_b64)
    iv = raw[:AES.block_size]
    ct = raw[AES.block_size:]
    cipher = AES.new(key, AES.MODE_CBC, iv)
    return unpad(cipher.decrypt(ct), AES.block_size).decode()

# ---------- RSA ----------
def rsa_encrypt(plain_text, public_key):
    cipher = PKCS1_OAEP.new(public_key)
    ct = cipher.encrypt(plain_text.encode())
    return base64.b64encode(ct).decode()

def rsa_decrypt(cipher_b64, private_key):
    cipher = PKCS1_OAEP.new(private_key)
    ct = base64.b64decode(cipher_b64)
    return cipher.decrypt(ct).decode()

# ---------- DEMO ----------
if __name__ == "__main__":
    # DES
    des_key = b'8bytekey'  # must be exactly 8 bytes
    des_ct = des_encrypt("Hello DES", des_key)
    des_pt = des_decrypt(des_ct, des_key)
    print("DES Encrypted:", des_ct)
    print("DES Decrypted:", des_pt)

    # AES
    aes_key = get_random_bytes(16)  # 16 bytes = AES-128
    aes_ct = aes_encrypt("Hello AES", aes_key)
    aes_pt = aes_decrypt(aes_ct, aes_key)
    print("AES Encrypted:", aes_ct)
    print("AES Decrypted:", aes_pt)

    # RSA
    key = RSA.generate(2048)
    pub_key = key.publickey()
    priv_key = key
    rsa_ct = rsa_encrypt("Hello RSA", pub_key)
    rsa_pt = rsa_decrypt(rsa_ct, priv_key)
    print("RSA Encrypted:", rsa_ct)
    print("RSA Decrypted:", rsa_pt)
```

```python
def caesar_encrypt(text, key):
    result = ''
    for ch in text:
        if ch.isupper():
            result += chr((ord(ch) - ord('A') + key) % 26 + ord('A'))
        elif ch.islower():
            result += chr((ord(ch) - ord('a') + key) % 26 + ord('a'))
        else:
            result += ch
    return result

def caesar_decrypt(cipher_text, key):
    return caesar_encrypt(cipher_text, -key)

if __name__ == "__main__":
    plaintext = "Hello Caesar!"
    key = 2

    encrypted = caesar_encrypt(plaintext, key)
    decrypted = caesar_decrypt(encrypted, key)

    print("Plaintext:", plaintext)
    print("Encrypted:", encrypted)
    print("Decrypted:", decrypted)
```
