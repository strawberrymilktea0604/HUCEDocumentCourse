```python
#Code không dùng OpenCV
#1. Xây dựng histogram

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
image_path = 'histogram.jpg'
image = Image.open(image_path)

# Chuyển đổi ảnh thành mảng numpy
image_array = np.array(image)
```

```python
colors = ('red', 'green', 'blue')
histograms = {}

for i, color in enumerate(colors):
    histograms[color] = np.histogram(image_array[:,:,i], bins=256, range=(0, 256))[0]
```

```python
plt.figure(figsize=(10, 6))

for color in colors:
    plt.plot(histograms[color], color=color)

plt.title('Histogram cho từng kênh màu')
plt.xlabel('Giá trị màu')
plt.ylabel('Tần suất')
plt.legend(colors)
plt.xlim([0, 255])
plt.show()
```

```python
#2. Cân bằng histogram

image = Image.open('histogram.jpg').convert('L')  # Chuyển ảnh sang đen trắng
image_array = np.array(image)


histogram, bin_edges = np.histogram(image_array, bins=256, range=(0, 256))

plt.figure(figsize=(10, 5))
plt.title("Histogram")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")
plt.bar(bin_edges[:-1], histogram, width=1, color='black')
plt.show()
```

```python
def histogram_equalization(image_array):
    histogram, _ = np.histogram(image_array.flatten(), bins=256, range=(0, 256))
    
    cdf = histogram.cumsum()  # Hàm phân phối tích lũy
    cdf_normalized = cdf * float(histogram.max()) / cdf.max()  # Chuẩn hóa CDF

    image_equalized = np.interp(image_array, bin_edges[:-1], cdf_normalized)
    return image_equalized

equalized_image = histogram_equalization(image_array)

plt.figure(figsize=(10, 5))
plt.title("Cân Bằng Histogram")
plt.imshow(equalized_image, cmap='gray')
plt.axis('off')
plt.show()
```
