```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter, laplace
from PIL import Image
import pandas as pd
import sys

# Hàm chuyển đổi ảnh RGB sang ảnh xám
def rgb2gray(image):
    return np.dot(image[...,:3], [0.2989, 0.5870, 0.1140])

# Hàm áp dụng Laplacian of Gaussian (LoG)
def laplacian_of_gaussian(image, sigma=1.0):
    # Bước 1: Làm mịn ảnh bằng Gaussian
    smoothed_image = gaussian_filter(image, sigma=sigma)
    
    # Bước 2: Tính Laplacian của ảnh đã làm mịn
    log_image = laplace(smoothed_image)
    
    # Chuẩn hóa về khoảng giá trị 0-255 để dễ hiển thị
    log_image = (log_image - log_image.min()) / (log_image.max() - log_image.min()) * 255
    
    return log_image

# Đọc ảnh từ file
image = Image.open('thuchanh.jpg')
image = np.array(image)

# Chuyển đổi ảnh thành ảnh xám
gray_image = rgb2gray(image)

# Áp dụng Laplacian of Gaussian
edges = laplacian_of_gaussian(gray_image, sigma=2)


pd.DataFrame(gray_image).to_csv('matrix1.csv', index=False)
pd.DataFrame(edges).to_csv('matrix2.csv', index=False)

# Hiển thị ảnh gốc và ảnh sau khi phát hiện cạnh
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(gray_image, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Laplacian of Gaussian (LoG)')
plt.imshow(edges, cmap='gray')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Đọc ảnh và chuyển thành ảnh xám
image = cv2.imread('thuchanh.jpg', cv2.IMREAD_GRAYSCALE)

# Bước 1: Làm mịn ảnh bằng Gaussian
sigma = 2  # Tham số điều chỉnh độ mịn
blurred_image = cv2.GaussianBlur(image, (0, 0), sigma)

# Bước 2: Áp dụng toán tử Laplacian để tính đạo hàm bậc hai
laplacian = cv2.Laplacian(blurred_image, cv2.CV_64F)

# Chuẩn hóa về khoảng giá trị 0-255 để hiển thị
laplacian = np.uint8(np.absolute(laplacian))


pd.DataFrame(image).to_csv('matrix3.csv', index=False)
pd.DataFrame(laplacian).to_csv('matrix4.csv', index=False)

# Hiển thị ảnh gốc và ảnh sau khi phát hiện cạnh
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(image, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Laplacian of Gaussian (LoG)')
plt.imshow(laplacian, cmap='gray')

plt.show()
```
