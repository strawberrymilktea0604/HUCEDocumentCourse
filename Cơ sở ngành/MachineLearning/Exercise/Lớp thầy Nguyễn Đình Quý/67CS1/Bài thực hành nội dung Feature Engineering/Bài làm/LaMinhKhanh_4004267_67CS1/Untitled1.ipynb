```python
#Code không thư viện
#Tải dữ liệu và xử lý dưới dạng danh sách

import csv
import math

# Tải dữ liệu Titanic từ file CSV
def load_data(file_path):
    with open(file_path, 'r') as file:
        reader = csv.reader(file)
        data = list(reader)
    return data

# Lấy dữ liệu Titanic
data = load_data('Housing.csv')

# Tách tiêu đề và dữ liệu
header = data[0]
data = data[1:]

# Hiển thị dữ liệu mẫu
for i in range(500):
    print(data[i])
```

```python
#1. Xử lý dữ liệu thiếu

def fill_missing_price(data, price_index):
    total_price = 0
    count = 0
    
    for row in data:
        if row[price_index] != '':
            total_price += float(row[price_index])
            count += 1
    
    average_price = total_price / count
    
    for row in data:
        if row[price_index] == '':
            row[price_index] = average_price
            

def fill_missing_area(data, area_index):
    total_area = 0
    count = 0
    
    for row in data:
        if row[area_index] != '':
            total_area += float(row[area_index])
            count += 1
    
    average_area = total_area / count
    
    for row in data:
        if row[area_index] == '':
            row[area_index] = average_area
            
def fill_missing_bedrooms(data, bedrooms_index):
    total_bedrooms = 0
    count = 0
    
    for row in data:
        if row[bedrooms_index] != '':
            total_bedrooms += float(row[bedrooms_index])
            count += 1
    
    average_bedrooms = total_bedrooms / count
    
    for row in data:
        if row[bedrooms_index] == '':
            row[bedrooms_index] = average_bedrooms
            

def fill_missing_bathrooms(data, bathrooms_index):
    total_bathrooms = 0
    count = 0
    
    for row in data:
        if row[bathrooms_index] != '':
            total_bathrooms += float(row[bathrooms_index])
            count += 1
    
    average_bathrooms = total_bathrooms / count
    
    for row in data:
        if row[bathrooms_index] == '':
            row[bathrooms_index] = average_bathrooms            

            
def fill_missing_stories(data, stories_index):
    total_stories = 0
    count = 0
    
    for row in data:
        if row[stories_index] != '':
            total_stories += float(row[stories_index])
            count += 1
    
    average_stories = total_stories / count
    
    for row in data:
        if row[stories_index] == '':
            row[stories_index] = average_stories
            
def fill_missing_parking(data, parking_index):
    total_parking = 0
    count = 0
    
    for row in data:
        if row[parking_index] != '':
            total_parking += float(row[parking_index])
            count += 1
    
    average_parking = total_parking / count
    
    for row in data:
        if row[parking_index] == '':
            row[parking_index] = average_parking
            


def fill_missing_mainroad(data, mainroad_index):
    mainroad_counts = {}
    
    for row in data:
        if row[mainroad_index] in mainroad_counts:
            mainroad_counts[row[mainroad_index]] += 1
        else:
            mainroad_counts[row[mainroad_index]] = 1
    
    mode_mainroad = max(mainroad_counts, key=mainroad_counts.get)
    
    for row in data:
        if row[mainroad_index] == '':
            row[mainroad_index] = mode_mainroad

def fill_missing_guestroom(data, guestroom_index):
    guestroom_counts = {}
    
    for row in data:
        if row[guestroom_index] in guestroom_counts:
            guestroom_counts[row[guestroom_index]] += 1
        else:
            guestroom_counts[row[guestroom_index]] = 1
    
    mode_guestroom = max(guestroom_counts, key=guestroom_counts.get)
    
    for row in data:
        if row[guestroom_index] == '':
            row[guestroom_index] = mode_guestroom            

def fill_missing_basement(data, basement_index):
    basement_counts = {}
    
    for row in data:
        if row[basement_index] in basement_counts:
            basement_counts[row[basement_index]] += 1
        else:
            basement_counts[row[basement_index]] = 1
    
    mode_basement = max(basement_counts, key=basement_counts.get)
    
    for row in data:
        if row[basement_index] == '':
            row[basement_index] = mode_basement  

            
def fill_missing_hotwaterheating(data, hotwaterheating_index):
    hotwaterheating_counts = {}
    
    for row in data:
        if row[hotwaterheating_index] in hotwaterheating_counts:
            hotwaterheating_counts[row[hotwaterheating_index]] += 1
        else:
            hotwaterheating_counts[row[hotwaterheating_index]] = 1
    
    mode_hotwaterheating = max(hotwaterheating_counts, key=hotwaterheating_counts.get)
    
    for row in data:
        if row[hotwaterheating_index] == '':
            row[hotwaterheating_index] = mode_hotwaterheating
            
def fill_missing_airconditioning(data, airconditioning_index):
    airconditioning_counts = {}
    
    for row in data:
        if row[airconditioning_index] in airconditioning_counts:
            airconditioning_counts[row[airconditioning_index]] += 1
        else:
            airconditioning_counts[row[airconditioning_index]] = 1
    
    mode_airconditioning = max(airconditioning_counts, key=airconditioning_counts.get)
    
    for row in data:
        if row[airconditioning_index] == '':
            row[airconditioning_index] = mode_airconditioning
            
def fill_missing_prefarea(data, prefarea_index):
    prefarea_counts = {}
    
    for row in data:
        if row[prefarea_index] in prefarea_counts:
            prefarea_counts[row[prefarea_index]] += 1
        else:
            prefarea_counts[row[prefarea_index]] = 1
    
    mode_prefarea = max(prefarea_counts, key=prefarea_counts.get)
    
    for row in data:
        if row[prefarea_index] == '':
            row[prefarea_index] = mode_prefarea

            
def fill_missing_furnishingstatus(data, furnishingstatus_index):
    furnishingstatus_counts = {}
    
    for row in data:
        if row[furnishingstatus_index] in furnishingstatus_counts:
            furnishingstatus_counts[row[furnishingstatus_index]] += 1
        else:
            furnishingstatus_counts[row[furnishingstatus_index]] = 1
    
    mode_furnishingstatus = max(furnishingstatus_counts, key=furnishingstatus_counts.get)
    
    for row in data:
        if row[furnishingstatus_index] == '':
            row[furnishingstatus_index] = mode_furnishingstatus

price_index = header.index('price')
area_index = header.index('area')
bedrooms_index = header.index('bedrooms')
bathrooms_index = header.index('bathrooms')
stories_index = header.index('stories')
mainroad_index = header.index('mainroad')
guestroom_index = header.index('guestroom')
basement_index = header.index('basement')
hotwaterheating_index = header.index('hotwaterheating')
airconditioning_index = header.index('airconditioning')
parking_index = header.index('parking')
prefarea_index = header.index('prefarea')
furnishingstatus_index = header.index('furnishingstatus')

fill_missing_price(data, price_index)
fill_missing_area(data, area_index)
fill_missing_bedrooms(data, bedrooms_index)
fill_missing_bathrooms(data, bathrooms_index)
fill_missing_stories(data, stories_index)
fill_missing_parking(data, parking_index)
fill_missing_mainroad(data, mainroad_index)
fill_missing_guestroom(data, guestroom_index)
fill_missing_basement(data, basement_index)
fill_missing_hotwaterheating(data, hotwaterheating_index)
fill_missing_airconditioning(data, airconditioning_index)
fill_missing_prefarea(data, prefarea_index)
fill_missing_furnishingstatus(data, furnishingstatus_index)
```

```python
#2. Chuẩn hóa dữ liệu

def standardize_column(data, index):
    values = [float(row[index]) for row in data]
    mean = sum(values) / len(values)
    std_dev = math.sqrt(sum([(x - mean) ** 2 for x in values]) / len(values))
    
    # Chuẩn hóa dữ liệu
    for row in data:
        row[index] = (float(row[index]) - mean) / std_dev


standardize_column(data, price_index)
standardize_column(data, area_index)
standardize_column(data, bedrooms_index)
standardize_column(data, bathrooms_index)
standardize_column(data, parking_index)
standardize_column(data, stories_index)
```

```python
#3. Mã hóa biến phân loại bằng One-Hot Encoding

def one_hot_encode(data, index):
    unique_values = list(set(row[index] for row in data))
    
    for value in unique_values:
        header.append(f"{header[index]}_{value}")
    
    for row in data:
        for value in unique_values:
            row.append(1 if row[index] == value else 0)
    for row in data:
        row.pop(index)
    header.pop(index)

one_hot_encode(data, mainroad_index)
one_hot_encode(data, guestroom_index)
one_hot_encode(data, basement_index)
one_hot_encode(data, hotwaterheating_index)
one_hot_encode(data, airconditioning_index)
one_hot_encode(data, prefarea_index)
one_hot_encode(data, furnishingstatus_index)
```
