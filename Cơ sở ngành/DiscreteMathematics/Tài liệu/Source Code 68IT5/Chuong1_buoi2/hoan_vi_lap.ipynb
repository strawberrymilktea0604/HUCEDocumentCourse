```python
def hoan_vi_lap(a, n):
    if(sum(n) == 0):
        return [[]]
    k = len(a)
    P = []
    for i in range(k):
        if(n[i] > 0):
            n_ = n.copy()
            n_[i] = n_[i] - 1
            for p in hoan_vi_lap(a, n_):
                p = [[a[i]]] + p
                P.append(p)
                
    return P       
```

```python
len(hoan_vi_lap(['a', 'b', 'c'], [4, 4, 3]))
```
