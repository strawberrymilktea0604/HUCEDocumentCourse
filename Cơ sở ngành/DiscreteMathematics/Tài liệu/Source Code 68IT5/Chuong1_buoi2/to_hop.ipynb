```python
def to_hop(a, k):
    n = len(a)
    if k == 1:
        return [[i] for i in a]
    if k == n:
        return [a]
    C = []
    b = a.copy()
    b.pop(0)
    for c in to_hop(b, k - 1):
        c = [a[0]] + c
        C.append(c)
    for c in to_hop(b, k):
        C.append(c)
    return C 
```

```python
for i in to_hop([1, 2, 3, 4, 5, 6, 7, 8, 9], 5):
    if(i[0] == 1 and i[1] == 3):
        print(i)
```

```python
#tinh so to hop
def C(n, k):
    if(k == 0 or n == k):
        return 1
    return C(n - 1, k - 1) + C(n - 1, k)
```
