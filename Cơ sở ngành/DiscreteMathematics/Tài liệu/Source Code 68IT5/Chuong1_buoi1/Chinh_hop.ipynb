```python
#Cach 2 (chua hoan thien)
def permutations_chinhhop(a, r):
    if r==1:
        return [[i] for i in a]
    P = []
    n = len(a)
    for i in range(n):
        b = a.copy()
        x = b.pop(i)
        for p in permutations_chinhhop(b, r-1):
            p = [x] + p
            P.append(p)
    return P
```

```python
#Tinh so chinh hop
def P(n, r):
    p = 1
    for i in range(r):
        p *= n- i
    return p
```
