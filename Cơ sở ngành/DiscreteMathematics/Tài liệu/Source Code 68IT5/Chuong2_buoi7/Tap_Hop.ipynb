```python
A = [1, 2, 3, 4, 3, 5, 4, 2, 1]
```

```python
for i in set(A):
    print(i, A.count(i))
```

```python
#Vi du ve co cua A
def power_set(A):
    if A == []:
        return[[]]
    P = [[]]
    for s in power_set(A[1:]):
        s = [A[0]] + s
        P.append(s)
    for s in power_set(A[1:]):
        if s != []:
            P.append(s)
    return P
```

```python
power_set([1, 2, 3])
```
