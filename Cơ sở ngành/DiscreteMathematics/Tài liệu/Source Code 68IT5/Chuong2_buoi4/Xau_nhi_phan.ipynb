```python
def xau_nhi_phan(n):
    if n == 1:
        return [[False], [True]]
    S = []
    for s in xau_nhi_phan(n - 1):
        s = [False] + s
        S.append(s)
    for s in xau_nhi_phan(n - 1):
        s = [True] + s 
        S.append(s)
    return S
```

```python
#VD3
#p == q Equivalent
# P = Equivalent(p, q)
# P = (p >> q) & (q >> p) 
P = (p & q) | ( (not p) & (not q) )
P
```

```python
for p_, q_ in xau_nhi_phan(2):
    print(P.subs({p : p_, q : q_}))
```

```python
p, q, r = symbols('p q r')
P = p |q & ~(r >> p)
P
```
