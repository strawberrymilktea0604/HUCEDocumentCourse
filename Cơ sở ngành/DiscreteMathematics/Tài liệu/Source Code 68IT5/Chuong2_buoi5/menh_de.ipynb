```python
from sympy import *
```

```python
p, q, r, s, t, u = symbols('p q r s t u')
```

```python
#Rut gon menh de
P = (~(p >> ~q)) & p
P.simplify()
```

```python
#Quy tac suy luan
#Modus Ponens(tách)
P = ((p >> q) & p) >> q
P
```

```python
P.simplify()
```

```python
#Cac quy tac lap luan
P = ((p >> q) & (q >> (r & s)) & (~r |~t | u) &(p & t)) >> u
P.simplify()
```
