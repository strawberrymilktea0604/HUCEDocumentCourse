```python
#Trong các xâu độ dài 3, theo thứ tự từ điển, xâu thứ 4 là?
#Đáp án: 011
def xau_nhi_phan(n):
    if n == 0:
        return [[]]
    else:
        P=[]
        for x in xau_nhi_phan(n-1):
            x=[0]+x
            P.append(x)
        for x in xau_nhi_phan(n-1):
            x=[1]+x
            P.append(x)    
        return P
```

```python
xau_nhi_phan(3)[3]
```

```python
# tìm xâu nhị phân đứng sau xâu [0,1,1]
def tim_xau(n):
    if xau_nhi_phan(n) == [[]]:
        return "Ko thoa man yeu cau"
    else:
        P = xau_nhi_phan(n)
        n=len(P) 
        for i in range(n):
             if P[i] == [0,1,1]:
                return P[i+1]
```

```python
tim_xau(3)
```

```python
#rong các hoán vị của 1, 2, 3, theo thứ tự từ điển, hoán vị thứ 2 là?
#dp án: 132
def tim_hoan_vi(a):
    n=len(a)
    if n == 0:
        return [[]]
    else:
        P=[]
        for i in range(n):
                b=a.copy()
                b.pop(i)
                for x in tim_hoan_vi(b):
                    x=[a[i]]+x
                    P.append(x)
        return P
```

```python
#Có bao nhiêu cách xếp 3 vật lấy ra từ 5 vật?
#Đáp án: 60
def xep(n, k):
    p = 1
    for i in range (n-k+1, n+1):
        p *= i
    return p 
```

```python
# Trong các chỉnh hợp chập 3 của 1, 2, 3, 4 theo thứ tự từ điển, chỉnh hợp thứ 5 là?
# Đáp án: 142
def xep_3_vat(a,k):
    if k==1:
        return [[x] for x in a]
    n=len(a)
    P=[]
    for i in range(n):
        b=a.copy()
        b.pop(i)
        for x in xep_3_vat(b,k-1):
            x=[a[i]]+x
            P.append(x)
    return P
```

```python
# có bao nhiêu cách chọn 3 vật từ 5vật
def chon(n, k):
    p = 1
    for i in range (1, k+1):
        p *= i
    q = 1
    for i in range (n-k+1, n+1):
        q *= i
    return q // p
```

```python
chon(5, 3)
```

```python
# Trong các tổ hợp chập 3 của 1, 2, 3, 4, 5, theo thứ tự từ điển, tổ hợp thứ 5 là?
# Đáp án: 135
def chon_m_vat(a,k):
    n=len(a)
    if k==1:
        return [[x] for x in a]
    if k==n:
        return [a]
    P=[]
    b=a.copy()
    b.pop(0)
    for x in chon_m_vat(b,k-1):
        x=[a[0]]+x
        P.append(x)
    for c in chon_m_vat(b,k):
        x=[a[0]]+c
        P.append(c)
    return P
```

```python
# Có bao nhiêu hoán vị (lặp) các chữ cái của từ COOL?
# Đáp án: 12
#Trong các hoán vị trên, theo thứ tự từ điển, từ xếp thứ 10 là?
#Đáp án: LOCO
def hoan_vi_lap(a,n):
    if sum(n)==0:
        return [[]]
    P=[]
    k=len(a)
    for i in range(k):
        if n[i] > 0:
            b=n.copy()
            b[i]-=1
            for x in hoan_vi_lap(a,b):
                x=[a[i]]+x
                P.append(x)
    return P
```

```python
# Có bao nhiêu cách chọn 4 vật (có lặp) từ 3 vật A, B, C?
# Đáp án: 15
# Trong các cách chọn trên, theo thứ tự từ điển, cách chọn thứ 5 là?
# Đáp án: AABB
def to_hop_lap(a,k):
    if k==1:
        return [[x] for x in a]
    n=len(a)
    if n==1:
        return [a*k]
    P=[]
    for x in to_hop_lap(a,k-1):
        x= [a[0]] + x
        P.append(x)
    for x in to_hop_lap(a[1:],k):
        P.append(x)
    return P
```

```python
#duong di
def duong_di(a,b,x,y):
    if b==y:
        return ['R'*(x-a)]
    if a==x:
        return ['U'*(y-b)]
    P=[]
    for p in duong_di(a+1,b,x,y):
        p='R'+p
        P.append(p)
    for p in duong_di(a,b+1,x,y):
        p='U'+p
        P.append(p)
    return P
```

```python
#13: Có bao nhiêu nghiệm nguyên không âm của phương trình x1 + x2 + x3 = 4?
# Đáp án: 15
def find_solutions(n, k):
    if k == 1:
        return [[n]]
    
    ds = []
    for i in range(n + 1):
        for con in find_solutions(n - i, k - 1):
            ds.append([i] + con)
    return ds

solutions  = find_solutions(4, 3)

len(solutions)
#solutions[4]
```

```python
#14: Có bao nhiêu tổng riêng của 10?
# Đáp án: 512
# Trong các tổng riêng trên, theo thứ tự từ điển, tổng riêng thứ 500 (viết liên tiếp các số hạng thành xâu) là?
#Đáp án: 613
# Trong các tổng riêng trên và có 4 số hạng, tổng riêng thứ 10 là?
def total(n, k):
    """Sinh các bộ số nguyên không âm có tổng = n và độ dài = k"""
    if k == 1:
        if n >= 1:
            return [[n]]
        else:
            return []
    result = []
    for i in range(1, n - k + 2):  # i >= 1
        for rest in total(n - i, k - 1):
            result.append([i] + rest)
    return result

# Gom tất cả tổng riêng của 10 (với số hạng từ 1 đến 10)
separate_total = []
for k in range(1, 11):
    separate_total += total(10, k)

separate_total.sort()

print(len(separate_total))

# In tổng riêng thứ 500 (chỉ số bắt đầu từ 0)
print(separate_total[499])

# Sinh riêng các tổng có 4 số hạng
separate_total_4 = total(10, 4)

# In tổng riêng thứ 10 trong nhóm có 4 số hạng
print(separate_total_4[9])
```

```python
# Viết giá trị chân lý của công thức mệnh đề P= p v q
#thành một xâu, theo thứ tự giảm dần của các bộ phân bố giá trị.
def gia_tri_chan_ly(n):
    if n == 1:
        return [[True], [False]]
    A = []
    for a in gia_tri_chan_ly(n - 1):
        a = [True] + a
        A.append(a)
    for a in gia_tri_chan_ly(n - 1):
        a = [False] + a
        A.append(a)
    return A
       
```

```python
# cos bao nhieu tap con cua tap co 10 phan tu
# Đáp án: 1024
def tap_con(n):
    if n == 0:
        return [[]]
    else:
        P = []
        for x in tap_con(n - 1):
            P.append(x)
            P.append([n] + x)
        return P
```

```python
# Trong các tập con của tập {1, 2, 3, 4}, theo thứ tự từ điển, tập thứ 8 (viết liên tiếp các phần tử thành xâu) là?
# Đáp án: 134
def tap_con (a: list):
    if len(a) == 0:
        return [[]]
    P = [[]]
    for s in tap_con(a[1:]):
        s = [a[0]] + s
        P.append(s)
    for s in tap_con(a[1:])[1:]:
        P.append(s)
    return P
```
