```python
a = []
```

```python
def res(a):
    n = len(a)
    a_i = 1
    for i in range(1, n - 1):
        e_j = 0
        for j in range(i):
            e_k = 0
            for k in range(i + 1, n):
                e_k = e_k or (a[i] == a[j] + a[k])
            e_j = e_j or e_k
        a_i = a_i and e_j
    return a_i
```

```python
import random
```

```python
a = random.sample(range(-10, 10, 1), 10)
a
```

```python
res(a)
```

```python
while True:
    a = random.sample(range(-10, 10, 1), 10)
    if(res(a)):
        break
a
```
