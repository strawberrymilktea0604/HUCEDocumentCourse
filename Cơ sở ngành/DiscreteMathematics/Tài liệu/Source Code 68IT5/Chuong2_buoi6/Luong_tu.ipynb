```python
#Vi du
U = [1, 4, 5]
V = [-1, 0, 2, 7]
```

```python
def P(x, y):
    return (x + y) % 2 == 0 or x < y
        
```

```python
#Bang gia tri P(x, y)

for x in U:
    for y in V:
        print(P(x, y), end=' ')
    print()
```

```python
#b, Xac dinh menh de
e = 0
for x in U:
    a = 1
    for y in V:
        a = a & P(x, y)
    e = e |a
    print(a, end= ' ')
e
```
