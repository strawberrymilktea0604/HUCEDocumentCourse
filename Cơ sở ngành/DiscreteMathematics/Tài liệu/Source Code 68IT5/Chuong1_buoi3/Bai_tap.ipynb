```python
from sympy import *
```

```python
n, i, j = symbols('n i j')
```

```python
#Bai 1
Sum(1, (j, i, n), (i, 1, n)).doit().simplify()
```

```python
#Bai 2
Sum(1, (k, j, n), (j, i, n), (i, 1, n)).doit().simplify().factor()
```
