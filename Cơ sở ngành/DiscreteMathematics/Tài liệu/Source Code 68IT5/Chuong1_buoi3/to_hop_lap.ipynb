```python
def to_hop_lap(a, k): # a = [A, B, C] k =4
    if(k == 1):
        return [[i] for i in a]
    n = len(a)
    if(n == 1):
        return [a * k]
    C = []
    for c in to_hop_lap(a, k -1):
        c = [a[0]] + c
        C.append(c)
    for c in to_hop_lap(a[1:], k):
        C.append(c)
    return C
```

```python
to_hop_lap(['A', 'B', 'C', 'D','E'], 5)[21]
```

```python
#dem
def dem_to_hop_lap(n, k):
    if(n == 1): 
        return 1
    if(k == 1):
        return n
    return dem_to_hop_lap(n, k - 1) + dem_to_hop_lap(n-1, k)
```

```python
dem_to_hop_lap(6, 5)
```

```python
#Dem so luong phan tu x1 + x2 + x3 + ... + xn = k
def solution(n, k):
    if n== 1:
        return [[k]]
    S = []
    for i in range(k + 1):
        for s in solution(n - 1, k - i):
            s = [i] + s
            S.append(s) 
        
    return S
```

```python
count = 0
for i in solution(3, 12):
    if(i[0] < 5):
        count += 1
count
```

```python
#
def summands(k):
    if(k == 1):
        return [[k]]
    S = [[k]]
    for i in range(k - 1, 0, -1):
        for s in summands(k - i):
            s = [i] + s 
            S.append(s)
    return S
        
```
