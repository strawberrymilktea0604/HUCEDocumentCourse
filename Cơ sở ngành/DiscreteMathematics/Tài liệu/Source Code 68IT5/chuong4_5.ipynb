```python
#Cho dãy (Fn) thỏa mãn F0 = 0, F1 = 1,và Fn = Fn − 1 + Fn − 2. Tính F10 
# Đáp án: 55
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

```python
#Tìm giá trị của biểu diễn 1402(5)
#đáp án: 227
def tinh_n(a : list, b : int) -> int:
    a.reverse()
    k = len(a) # hơn k lý thuyết 1 đơn vị
    n = a[k-1]
    for i in range(k-2, -1, -1):
        n = b * n + a[i]
    return n
```

```python
tinh_n([1, 4, 0, 2], 5)
```

```python
#Tìm biểu diễn của 1966 trong cơ số 5
#Đáp án: (30331)5
def tim_bieu_dien(n : int, b : int) -> list:
    a =[]
    while n != 0:
        a.append(n % b)
        n = n // b
    a.reverse()
    return a
```

```python
#Phân tích 1960 thành thừa số nguyên tố
#Đáp án: 2^3⋅5⋅7^2
def phan_tich_thua_so(n : int) -> dict:
    i = 2
    res = {}
    while n != 1:
        while n % i !=0:
            i += 1
        res[i] = 0
        while n % i == 0:
            n = n // i
            res[i] += 1
    return res
```

```python
'''a = [2, 4, 3]    # Số 342 trong hệ cơ số 5
b = [3, 2, 1, 4] # Số 4123 trong hệ cơ số 5
base = 5         # Cơ số 5
k = len(a)       # Độ dài mảng a
l = len(b)       # Độ dài mảng b
s = [0] * (k + l) # Mảng kết quả (đủ chỗ cho tích)

for i in range(l):
    r = R = 0
    for j in range(k):
        t = a[j] * b[i] + r  # Nhân chữ số + nhớ
        p = t % base          # Phần dư
        r = t // base         # Số nhớ
        t = s[i+j] + p + R   # Cộng vào kết quả hiện tại
        s[i+j] = t % base     # Lưu phần dư
        R = t // base         # Cập nhật nhớ
    s[k + i] = r + R          # Lưu nhớ cuối cùng

# Kết quả: s = [1, 1, 0, 3, 4, 3, 1] → 1343011 trong hệ cơ số 5'''
#viết đoạn code trên thành hàm
def nhan_hai_so(a : list, b : list, base : int) -> list:
    k = len(a)
    l = len(b)
    s = [0] * (k + l)
    for i in range(l):
        r = R = 0
        for j in range(k):
            t = a[j] * b[i] + r
            p = t % base
            r = t // base
            t = s[i+j] + p + R
            s[i+j] = t % base
            R = t // base
        s[k + i] = r + R
    s.reverse()
    return s
```

```python
#Số 1960 có bao nhiêu ước nguyên dương?
#Đáp án: 24
def uoc_nguyen(n : int) -> int:
    res = 1
    a = phan_tich_thua_so(n)
    for i in a:
        res *= (a[i] + 1)
    return res
#giai thích: nếu n = p1^a1 * p2^a2 * ... * pk^ak thì số ước nguyên dương của n là (a1 + 1)(a2 + 1)...(ak + 1)
```

```python
#tìm ước chung lớn nhất của 2 số
def ucln(a, b):
    if b == 0:
        return a
    return ucln(b, a % b)
```

```python
'''Tìm ước chung lớn nhất (GCD) của 2 số a và b
  Ví dụ: gcdex(287, 91) trả về GCD là 7
Tìm 2 số nguyên x, y thỏa mãn phương trình Bézout:
    a*x + b*y = GCD(a,b)
Ví dụ: Với a=287, b=91, hàm trả về x=-3, y=10 vì:
287*(-3) + 91*10 = -861 + 910 = 49 (Ở đây có sai số, GCD thực tế là 7)'''
def gcdex(a, b):
    x0, y0 = 1, 0
    x1, y1 = 0, 1
    while b != 0:
        q = a // b
        a, b = b, a % b
        x = x0 - x1 * q
        y = y0 - y1 * q
        x0, y0 = x1, y1
        x1, y1 = x, y
    return x0, y0, a
```

```python
gcdex(287, 91)
```

```python
#CHƯƠNG 5: HÀM
#Có bao nhiêu quan hệ từ tập cỡ 2 vào tập cỡ 3?
#Đáp án: 64
def so_luong_quan_he(so_phan_tu_A: int, so_phan_tu_B: int) -> int:
    """
    Tính số quan hệ từ tập A (cỡ m) vào tập B (cỡ n).
    Công thức: 2^(m * n).
    """
    return 2 ** (so_phan_tu_A * so_phan_tu_B)
```

```python
so_luong_quan_he(2, 3)
```

```python
#Có bao nhiêu hàm từ tập cỡ 4 vào tập cỡ 3 (số cách xếp 4 vật vào 3 hộp)?
#Đáp án: 81
def so_luong_ham(so_phan_tu_A: int, so_phan_tu_B: int) -> int:
    """
    Tính số hàm từ tập A (cỡ m) vào tập B (cỡ n).
    Công thức: n^m.
    """
    return so_phan_tu_B ** so_phan_tu_A
```

```python
so_luong_ham(4, 3)
```

```python
#Có bao nhiêu đơn ánh từ tập cỡ 3 vào tập cỡ 5 (số cách xếp 3 vật vào 5 hộp sao cho mỗi hộp không quá 1 vật)?
#Đáp án: 60
def so_luong_don_anh(so_phan_tu_A: int, so_phan_tu_B: int) -> int:
    """
    Tính số đơn ánh từ tập A (cỡ m) vào tập B (cỡ n).
    Công thức:P(n, m) = n! / (n - m)!.
    """
    return factorial(so_phan_tu_B) // factorial(so_phan_tu_B - so_phan_tu_A)
```

```python
# tính giá trị của xâu nhị phân 11010100
# Đáp án: 212
def tinh_nhi_phan(a: str) -> int:
    n = 0
    for i in range(len(a)):
        n = n * 2 + int(a[i])
    return n
```

```python
tinh_nhi_phan("11010100")
```

```python
# biểu diễn n = 40545 trong hệ bát phân
# Đáp án: 124541
def bcd(n: int) -> str:
    if n == 0:
        return "0"
    s = ""
    while n > 0:
        s = str(n % 8) + s
        n //= 8
    return s
```
